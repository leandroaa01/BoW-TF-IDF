```python
import pandas as pd
from numpy import mean, std

## Carregando o dataset
dataset = pd.read_csv('Imdb_TF-IDF.csv', encoding='utf-8')

## Features (tudo menos a última coluna) e target (última coluna)
X = dataset.iloc[:, :-1].values       # .values ajuda a MLP
y = dataset.iloc[:, -1].values        # funciona independente do nome


## Algoritmo
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import cross_val_score, KFold

kf = KFold(n_splits=5, random_state=1, shuffle=True)

mlp = MLPClassifier(hidden_layer_sizes=(100,), activation='relu',
                    solver='adam', max_iter=500, random_state=1)

scores = cross_val_score(mlp, X, y, scoring='accuracy', cv=kf)
print('Accuracy: %.3f (%.3f)' % (mean(scores), std(scores)))
```

```python
## Utilizando uma métrica de tempo
import time

### Importing Dataset
dataset = pd.read_csv('HOG_64_20x20_PCA-75.csv',encoding='utf-8')

## Selecionando todos os atributos com exceção da classe
dataset_part01 = dataset.iloc[:, :-1]

X = dataset_part01      # Features
y = dataset.classe      # Target variable (classe)

## Carregando o algoritmo
from sklearn.neural_network import MLPClassifier
from sklearn import metrics

# Separando dataset em duas partes: treinamento e teste
from sklearn.model_selection import train_test_split

## Montando um GridSearch
par_hls = [(100), (120), (150), (50,50), (60,60)]
par_activation = ['identity', 'logistic', 'relu', 'tanh']
par_solver = ['adam', 'sgd']
par_learning_rate_init = [0.0001, 0.001, 0.01]
par_max_iter = [500, 1000, 1500, 2000, 2500]

### Começando a contar o tempo
start_time = time.time()

for i in par_hls:
    for j in par_activation:
        for k in par_solver:
            for l in par_learning_rate_init:
                for m in par_max_iter:
                    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=19 )

                    mlp = MLPClassifier(hidden_layer_sizes=i, activation=j,
                                        solver=k, learning_rate_init=l, max_iter=m)

                    mlp.fit(X_train, y_train)
                    y_pred = mlp.predict(X_test)

                    accuracy = metrics.accuracy_score(y_test, y_pred)
                    print('Accuracy: %.3f, hidden: %s, activation = %s, solver = %s, learning = %s, max_iter = %s'
                     % (accuracy, i, j, k, l, m))

end_time = time.time()
# Tempo total de treinamento
execution_time = end_time - start_time
print(f"Tempo de execução para treinamento: {execution_time:.4f} segundos")
```

```python
import pandas as pd
from numpy import mean
from numpy import std
import warnings

warnings.filterwarnings("ignore")

## Utilizando uma métrica de tempo
import time

### Importing Dataset
dataset = pd.read_csv('HOG_64_20x20_PCA-75.csv',encoding='utf-8')

## Selecionando todos os atributos com exceção da classe
dataset_part01 = dataset.iloc[:, :-1]

X = dataset_part01      # Features
y = dataset.classe      # Target variable (classe)

## Carregando o algoritmo
from sklearn.neural_network import MLPClassifier
from sklearn import metrics

## Implementando k-fold CV
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import cross_val_predict
from sklearn.model_selection import KFold

### Configurações
configs = []

### Montando um GridSearch
par_hls = [(100), (120), (150), (50,50), (60,60)]
par_activation = ['identity', 'logistic', 'relu', 'tanh']
par_solver = ['adam', 'sgd']
par_learning_rate_init = [0.0001, 0.001, 0.01]
par_max_iter = [500, 1000, 1500, 2000]

### Começando a contar o tempo
start_time = time.time()

for i in par_hls:
    for j in par_activation:
        for k in par_solver:
            for l in par_learning_rate_init:
                for m in par_max_iter:
                  kf = KFold(n_splits=5, random_state=1, shuffle=True)

                  mlp = MLPClassifier(hidden_layer_sizes=i, activation=j,
                                        solver=k, learning_rate_init=l, max_iter=m)

                  # Model Accuracy
                  scores = cross_val_score(mlp, X, y, scoring='accuracy', cv=kf)

                  ### Inserir no configs
                  config = {
                      "Accuracy": mean(scores),
                      "Hidden": i,
                      "Activation": j,
                      "Solver": k,
                      "Learning": l,
                      "Max_iter": m
                  }

                  # Adiciona o dicionário à lista principal
                  configs.append(config)

end_time = time.time()
# Tempo total de treinamento
execution_time = end_time - start_time
print(f"Tempo de execução para treinamento: {execution_time:.4f} segundos")

df_resultados = pd.DataFrame(configs)
df_resultados = df_resultados.sort_values(by="Accuracy", ascending=False).head(15)

## Printando ...
print('Base: HOG_64_20x20_PCA-75.csv')
print(df_resultados)
```
